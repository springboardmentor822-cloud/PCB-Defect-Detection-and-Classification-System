```python
import numpy as np
import pandas as pd
import re

from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer

from gensim import corpora
from gensim.models import LdaModel

import nltk
nltk.download('vader_lexicon')
from nltk.sentiment import SentimentIntensityAnalyzer

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
newsgroups = fetch_20newsgroups(
    subset='all',
    remove=('headers','footers','quotes')
)

texts = newsgroups.data
```

```python
import re
import nltk
nltk.download('stopwords')

from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    words = text.split()
    words = [w for w in words if w not in stop_words and len(w) > 3]
    return words

processed_texts = [preprocess(doc) for doc in texts]
```

```python
from gensim import corpora

dictionary = corpora.Dictionary(processed_texts)
```

```python
dictionary.filter_extremes(no_below=15, no_above=0.5)
```

```python
corpus = [dictionary.doc2bow(text) for text in processed_texts]
```

```python
from gensim.models import LdaModel

lda_model = LdaModel(
    corpus=corpus,
    id2word=dictionary,
    num_topics=30,
    passes=10,
    random_state=42
)
```

```python
for i in range(30):
    print(f"\nTopic {i}:")
    print(lda_model.show_topic(i, topn=10))
```

```python
topic_labels = {}

def label_topic(words):
    words = [w for w, _ in words]

    if any(w in words for w in ['game','team','hockey','season','player']):
        return "Sports"

    elif any(w in words for w in ['jesus','god','bible','christian','church']):
        return "Religion"

    elif any(w in words for w in ['government','law','israel','state','political']):
        return "Politics"

    elif any(w in words for w in ['windows','file','program','software','computer']):
        return "Software/OS"

    elif any(w in words for w in ['disk','drive','scsi','hardware','card']):
        return "Hardware"

    elif any(w in words for w in ['encryption','security','key','chip','data']):
        return "Cybersecurity"

    elif any(w in words for w in ['email','internet','ftp','mail','network']):
        return "Internet/Communication"

    elif any(w in words for w in ['price','sale','car','buy','new']):
        return "Marketplace"

    elif any(w in words for w in ['help','thanks','anyone','please','know']):
        return "General Query"

    else:
        return "General/Misc"
```

```python
for i in range(30):
    words = lda_model.show_topic(i, topn=10)
    label = label_topic(words)

    print(f"\nTopic {i} → {label}")
    print(words)
```

```python
predicted_topics = []

for doc in corpus:
    topics = lda_model.get_document_topics(doc)
    dominant_topic = max(topics, key=lambda x: x[1])[0]
    predicted_topics.append(dominant_topic)
```

```python
true_labels = newsgroups.target
```

```python
from sklearn.metrics import normalized_mutual_info_score

nmi_score = normalized_mutual_info_score(true_labels, predicted_topics)

print("NMI Score (LDA):", nmi_score)
```

```python
from nltk.sentiment import SentimentIntensityAnalyzer

sia = SentimentIntensityAnalyzer()

def analyze_sentiment(text):
    scores = sia.polarity_scores(text)
    compound = scores['compound']

    # Slightly stricter thresholds (unique tweak)
    if compound >= 0.1:
        return "Positive"
    elif compound <= -0.1:
        return "Negative"
    else:
        return "Neutral"
```

```python
sentiment_results = list(map(analyze_sentiment, texts))
```

```python
dataset = pd.DataFrame({
    "text": texts,
    "topic": predicted_topics,
    "sentiment": sentiment_results
})
```

```python
topic_sentiment = dataset.groupby(['topic', 'sentiment']).size().unstack().fillna(0)

print(topic_sentiment)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
sns.heatmap(topic_sentiment, annot=True, fmt=".0f", cmap="coolwarm")
plt.title("Topic vs Sentiment Distribution")
plt.xlabel("Sentiment")
plt.ylabel("Topic")
plt.show()
```

```python
topic_summary = []

for topic_id in range(30):
    # Get top words
    words = lda_model.show_topic(topic_id, topn=10)
    top_words = ", ".join([w for w, _ in words])

    # Get sentiment distribution for this topic
    topic_data = dataset[dataset['topic'] == topic_id]['sentiment']
    sentiment_counts = topic_data.value_counts()

    # Get dominant sentiment
    if len(sentiment_counts) > 0:
        dominant_sentiment = sentiment_counts.idxmax()
    else:
        dominant_sentiment = "No Data"

    topic_summary.append({
        "Topic": topic_id,
        "Top Words": top_words,
        "Dominant Sentiment": dominant_sentiment
    })

summary_df = pd.DataFrame(topic_summary)
summary_df
```

```python
summary_df.sort_values(by="Dominant Sentiment")
```

```python
plt.figure(figsize=(12,6))
sns.countplot(x="Dominant Sentiment", data=summary_df)
plt.title("Dominant Sentiment Across Topics")
plt.show()
```

```python
def extractive_summary(text, num_sentences=2):
    # Skip very short text
    if len(text.split()) < 50:
        return text

    try:
        parser = PlaintextParser.from_string(text, Tokenizer("english"))
        summarizer = LsaSummarizer()

        summary = summarizer(parser.document, num_sentences)
        return " ".join([str(sentence) for sentence in summary])

    except:
        return "Summary not available"
```
